```python
import time
import random
```

```python
data = []
start_timestamp = int(time.mktime(time.strptime("2023-08-01", "%Y-%m-%d")))
end_timestamp = int(time.mktime(time.strptime("2023-08-10", "%Y-%m-%d")))
day_difference = 24 * 60 * 60  # 1 day in seconds

for timestamp in range(start_timestamp, end_timestamp, day_difference):
    timestamp = timestamp*1000 #for miliseconds formating
    theft = {"timestamp":timestamp,"count":random.randint(1,9),"event_name":"theft","context":"in"}
    data.append(theft)
    loitering = {"timestamp":timestamp,"count":random.randint(1,9),"event_name":"loitering","context":"in"}
    data.append(loitering)
    car_mov_in = {"timestamp":timestamp,"count":random.randint(1,9),"event_name":"car_mov","context":"in"}
    data.append(car_mov_in)
    car_mov_out = {"timestamp":timestamp,"count":random.randint(1,9),"event_name":"car_mov","context":"out"}
    data.append(car_mov_out)
    people_mov_in = {"timestamp":timestamp,"count":random.randint(1,9),"event_name":"people_mov","context":"in"}
    data.append(people_mov_in)
    people_mov_out = {"timestamp":timestamp,"count":random.randint(1,9),"event_name":"people_mov","context":"out"}
    data.append(people_mov_out)
    
data
```

```python
import json

# Specify the file path where you want to save the JSON data
file_path = "movements_script.json"

# Write the dictionary to a JSON file
with open(file_path, "w") as json_file:
    json.dump(data, json_file, indent=4)
```
